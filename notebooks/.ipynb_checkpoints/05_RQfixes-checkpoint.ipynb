Loads all pre-computed results from `results/`. Does not rerun the main LOSO loop.

```python
import numpy as np
import pickle
import warnings
from pathlib import Path
from scipy.stats import wilcoxon, spearmanr
%pip install statsmodels
from statsmodels.stats.multitest import multipletests
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

try:
    from scipy.stats import binomtest
    def binom_pval(k, n, alt='greater'):
        return binomtest(k, n, 0.5, alternative=alt).pvalue
except ImportError:
    from scipy.stats import binom_test
    def binom_pval(k, n, alt='greater'):
        return binom_test(k, n, 0.5, alternative=alt)

# Resolve root regardless of where notebook is launched from
ROOT = Path('..') if Path('../results').exists() else Path('.')
RES  = ROOT / 'results'
FIGS = RES / 'figures'
FIGS.mkdir(exist_ok=True)

X    = np.load(RES / 'X_features.npy')       # (1280, 131): 128 log-band-power + 3 FAA
y    = np.load(RES / 'y.npy')                 # (1280, 4): valence, arousal, dominance, liking
sids = np.load(RES / 'subject_ids.npy')       # (1280,): integer subject IDs

with open(RES / 'loso_results.pkl',   'rb') as f: loso = pickle.load(f)
with open(RES / 'within_subject.pkl', 'rb') as f: ws   = pickle.load(f)

# Pre-computed 131-dim importance vectors (mean across subjects/folds)
imp_rf  = {t: np.load(RES / f'importance_mean_{t}.npy')     for t in ['valence', 'arousal', 'dominance']}
imp_svm = {t: np.load(RES / f'svm_importance_mean_{t}.npy') for t in ['valence', 'arousal', 'dominance']}

SUBJECTS = np.unique(sids)
BANDS    = {'theta': 0, 'alpha': 1, 'beta': 2, 'gamma': 3}
T_IDX    = {'valence': 0, 'arousal': 1, 'dominance': 2}

print(f'X: {X.shape}  |  subjects: {len(SUBJECTS)}')
print('loso keys:', list(loso.keys()) if isinstance(loso, dict) else type(loso))
print('ws   keys:', list(ws.keys())   if isinstance(ws,   dict) else type(ws))
```

```python
# Shared helpers used across all RQ sections

def binarize(y_raw, thresh):
    """Binary split: trials strictly above threshold -> 1, at or below -> 0."""
    return (y_raw > thresh).astype(int)

def band_feat(X_full, band_idx):
    """Extract 32-channel features for one band from the 131-feature matrix.
    Layout is channel-major: ch0_theta, ch0_alpha, ch0_beta, ch0_gamma, ch1_theta, ...
    So band_idx in {0,1,2,3} maps to theta/alpha/beta/gamma respectively.
    Slicing stops at 128 to exclude FAA features at indices 128-130.
    """
    return X_full[:, band_idx:128:4]   # shape (n_trials, 32)

def loso_rf_acc(X_b, y_col, n_est=50):
    """LOSO RF mean accuracy for a given feature matrix and raw label column.
    Uses fold-specific median split and StandardScaler to match original pipeline.
    """
    accs = []
    for s in SUBJECTS:
        te = sids == s
        tr = ~te
        Xtr, Xte   = X_b[tr], X_b[te]
        ytr_r, yte_r = y_col[tr], y_col[te]
        thresh = np.median(ytr_r)
        ytr, yte = binarize(ytr_r, thresh), binarize(yte_r, thresh)
        if len(np.unique(ytr)) < 2:
            continue
        sc = StandardScaler()
        Xtr = sc.fit_transform(Xtr)
        Xte = sc.transform(Xte)
        rf  = RandomForestClassifier(n_estimators=n_est, random_state=42, n_jobs=1)
        rf.fit(Xtr, ytr)
        accs.append(rf.score(Xte, yte))
    return np.mean(accs)

def safe_extract(d, *keys):
    """Navigate nested dict by key sequence; return None on any miss."""
    for k in keys:
        if not isinstance(d, dict) or k not in d:
            return None
        d = d[k]
    return d

print('Helpers defined.')
```

## RQ1: Matched Per-Band Null Distributions

**Problem in original analysis (Limitation L10):** The shuffled-label null was estimated for the full 131-feature LOSO RF model. A 32-feature single-band model has tighter null variance than the full model, so the full-model null p95 (0.534) is not applicable to ablation results. Gamma-only (0.543) and beta-only (0.537) outperforming the full model for valence is directional without this fix.

**Fix:** For each of 4 bands and 3 targets, run 200 permutations of LOSO RF using only the 32-feature subset for that band with shuffled labels. This produces a matched null distribution with the correct variance for a 32-feature space.

**Formal test:** one-sided, observed accuracy vs matched null p95. If observed > p95: reject H0 (chance) for that band-target pair.

```python
# Observed band ablation accuracy (hardcoded from paper; overridden if found in loso_results.pkl)
OBS_ABL = {
    'theta': {'valence': 0.495, 'arousal': 0.499, 'dominance': 0.512},
    'alpha': {'valence': 0.504, 'arousal': 0.506, 'dominance': 0.472},
    'beta':  {'valence': 0.537, 'arousal': 0.514, 'dominance': 0.520},
    'gamma': {'valence': 0.543, 'arousal': 0.500, 'dominance': 0.502},
}

# Try loading from loso_results.pkl
for key in ['band_ablation', 'ablation', 'band_ablation_acc', 'ablation_acc']:
    candidate = safe_extract(loso, key)
    if candidate is not None:
        OBS_ABL = candidate
        print(f"Loaded ablation from loso_results['{key}']")
        break
else:
    print('Using hardcoded ablation values from summary.md (paper Table)')

# Compute matched per-band null distributions (checkpointed per band-target pair)
NULL_DIR = RES / 'rq1_band_null'
NULL_DIR.mkdir(exist_ok=True)
N_PERM = 200
rng_rq1 = np.random.default_rng(0)

band_nulls = {}
for bname, bidx in BANDS.items():
    band_nulls[bname] = {}
    Xb = band_feat(X, bidx)   # (1280, 32); reused across targets
    for tname, tidx in T_IDX.items():
        fpath = NULL_DIR / f'{bname}_{tname}.npy'
        if fpath.exists():
            band_nulls[bname][tname] = np.load(fpath)
            print(f'Loaded null: {bname}-{tname}')
        else:
            print(f'Computing null: {bname}-{tname} ({N_PERM} perms)...', end=' ', flush=True)
            y_col = y[:, tidx]
            nulls = np.array([loso_rf_acc(Xb, rng_rq1.permutation(y_col)) for _ in range(N_PERM)])
            np.save(fpath, nulls)
            band_nulls[bname][tname] = nulls
            print(f'p95={np.percentile(nulls, 95):.3f}')
```

```python
# Formal test: observed ablation accuracy vs matched null p95
print(f'RQ1: Band Ablation vs Matched Null  (n_perm={N_PERM}, LOSO RF, EEG-only)')
print(f'{"Band":<8} {"Target":<12} {"Obs":>8} {"Null p50":>10} {"Null p95":>10} {"p-val":>8}  Decision')
print('-' * 68)

rq1_results = {}
for bname in BANDS:
    for tname in T_IDX:
        obs  = np.mean(OBS_ABL[tname][bname]['acc'])
        null = band_nulls[bname][tname]
        p50  = np.percentile(null, 50)
        p95  = np.percentile(null, 95)
        # One-sided p-value: fraction of null >= observed (minimum = 1/N_PERM)
        pval = max(np.mean(null >= obs), 1.0 / N_PERM)
        rej  = obs > p95
        rq1_results.setdefault(bname, {})[tname] = {'obs': obs, 'p50': p50, 'p95': p95,
                                                      'pval': pval, 'reject': rej}
        dec = 'REJECT H0' if rej else 'fail to reject'
        print(f'{bname:<8} {tname:<12} {obs:>8.3f} {p50:>10.3f} {p95:>10.3f} {pval:>8.4f}  {dec}')

# Heatmap: observed - null p95 (green = formally above null)
blist = list(BANDS.keys())
tlist = list(T_IDX.keys())
diff_mat = np.array([[rq1_results[b][t]['obs'] - rq1_results[b][t]['p95']
                       for t in tlist] for b in blist])
obs_mat  = np.array([[rq1_results[b][t]['obs'] for t in tlist] for b in blist])

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

im0 = axes[0].imshow(obs_mat, cmap='YlOrRd', aspect='auto', vmin=0.46, vmax=0.56)
axes[0].set_xticks(range(3)); axes[0].set_xticklabels(tlist)
axes[0].set_yticks(range(4)); axes[0].set_yticklabels(blist)
axes[0].set_title('Observed LOSO RF accuracy (single-band models)')
for i in range(4):
    for j in range(3):
        axes[0].text(j, i, f'{obs_mat[i,j]:.3f}', ha='center', va='center', fontsize=9)
plt.colorbar(im0, ax=axes[0])

im1 = axes[1].imshow(diff_mat, cmap='RdYlGn', aspect='auto', vmin=-0.05, vmax=0.05)
axes[1].set_xticks(range(3)); axes[1].set_xticklabels(tlist)
axes[1].set_yticks(range(4)); axes[1].set_yticklabels(blist)
axes[1].set_title('Obs minus matched null p95  (green = formal above-chance)')
for i in range(4):
    for j in range(3):
        col = 'white' if abs(diff_mat[i,j]) > 0.03 else 'black'
        axes[1].text(j, i, f'{diff_mat[i,j]:+.3f}', ha='center', va='center',
                     fontsize=9, color=col)
plt.colorbar(im1, ax=axes[1])

plt.suptitle('RQ1: Matched Per-Band Null Comparison', fontsize=12, y=1.01)
plt.tight_layout()
plt.savefig(FIGS / 'rq1_band_null_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nConclusion:')
any_reject = False
for bname in BANDS:
    for tname in T_IDX:
        r = rq1_results[bname][tname]
        if r['reject']:
            any_reject = True
            print(f'  {bname}-only model is formally above chance for {tname} '
                  f'(obs={r["obs"]:.3f}, p95={r["p95"]:.3f}, p={r["pval"]:.4f})')
if not any_reject:
    print('  No band-target pair formally exceeds its matched null. All results remain directional.')
```

## RQ2: FAA-Only Within-Subject SVM

**Problem in original analysis:** FAA features (indices 128-130) competed with 128 band-power features in the full 131-feature model. Permuting one FAA feature does not fully remove its information because the classifier substitutes unpermuted individual alpha channels (compensatory structure), producing systematic downward bias on FAA importance. The within-subject test set of approximately 8 trials per fold is insufficient to detect a subtle effect in a high-dimensional competition.

**Fix:** Run a separate within-subject 5-fold SVM with only the 3 FAA features. This removes competition effects entirely and eliminates the compensatory structure bias.

**Formal tests:**
1. Binomial sign test per target: is the fraction of subjects with accuracy > 0.5 above chance?
2. Paired Wilcoxon across subjects: is valence FAA-only accuracy higher than arousal FAA-only accuracy, consistent with Davidson (1992)?

```python
X_faa = X[:, 128:131]   # (1280, 3): FAA features only

def faa_ws_cv(X_faa_all, y_col):
    """Within-subject 5-fold SVM accuracy using 3 FAA features. Returns per-subject array."""
    subj_accs = []
    for s in SUBJECTS:
        mask = sids == s
        Xs, yr = X_faa_all[mask], y_col[mask]
        thresh  = np.median(yr)
        ys      = binarize(yr, thresh)
        if len(np.unique(ys)) < 2:
            subj_accs.append(np.nan)
            continue
        skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
        fold_accs = []
        for tri, tei in skf.split(Xs, ys):
            Xtr, Xte = Xs[tri], Xs[tei]
            ytr, yte = ys[tri], ys[tei]
            if len(np.unique(ytr)) < 2:
                continue
            sc  = StandardScaler()
            Xtr = sc.fit_transform(Xtr)
            Xte = sc.transform(Xte)
            svm = SVC(kernel='rbf', C=10)
            svm.fit(Xtr, ytr)
            fold_accs.append(svm.score(Xte, yte))
        subj_accs.append(np.nanmean(fold_accs) if fold_accs else np.nan)
    return np.array(subj_accs)

faa_accs = {t: faa_ws_cv(X_faa, y[:, T_IDX[t]]) for t in T_IDX}

print('FAA-only within-subject SVM (3 features, C=10 RBF):')
for t, accs in faa_accs.items():
    valid = accs[~np.isnan(accs)]
    print(f'  {t:<12}: mean={np.nanmean(valid):.3f}  n_valid={len(valid)}')
```

```python
# Binomial sign test per target
print('RQ2: FAA-only Sign Test (H0: P(acc > 0.5) = 0.5, one-sided)')
for t, accs in faa_accs.items():
    valid  = accs[~np.isnan(accs)]
    n_ab   = int(np.sum(valid > 0.5))
    n_v    = len(valid)
    pval   = binom_pval(n_ab, n_v, alt='greater')
    sig    = '*' if pval < 0.05 else ''
    print(f'  {t:<12}: {n_ab}/{n_v} above 0.5  mean={np.nanmean(valid):.3f}  p={pval:.4f} {sig}')

# Paired Wilcoxon: valence FAA-only > arousal FAA-only (Davidson 1992 prediction)
val_a = faa_accs['valence']
aro_a = faa_accs['arousal']
dom_a = faa_accs['dominance']
shared_va = ~(np.isnan(val_a) | np.isnan(aro_a))
shared_vd = ~(np.isnan(val_a) | np.isnan(dom_a))

stat_va, p_va = wilcoxon(val_a[shared_va], aro_a[shared_va], alternative='greater')
stat_vd, p_vd = wilcoxon(val_a[shared_vd], dom_a[shared_vd], alternative='greater')
_, pvals_fdr, _, _ = multipletests([p_va, p_vd], method='fdr_bh')

print('\nWilcoxon: valence FAA-only acc > other target FAA-only acc (BH-FDR corrected)')
print(f'  val > aro: stat={stat_va:.2f}  p={p_va:.4f}  p_fdr={pvals_fdr[0]:.4f}  '
      f'{"REJECT H0" if pvals_fdr[0] < 0.05 else "fail to reject"}')
print(f'  val > dom: stat={stat_vd:.2f}  p={p_vd:.4f}  p_fdr={pvals_fdr[1]:.4f}  '
      f'{"REJECT H0" if pvals_fdr[1] < 0.05 else "fail to reject"}')

print('\nInterpretation:')
val_sign = binom_pval(int(np.sum(val_a[~np.isnan(val_a)] > 0.5)), len(val_a[~np.isnan(val_a)]))
if val_sign < 0.05 and pvals_fdr[0] < 0.05:
    print('  FAA is formally above chance for valence AND specifically more discriminative '
          'for valence than arousal. Consistent with Davidson (1992). RQ2: REJECT null.')
elif val_sign < 0.05:
    print('  FAA is above chance for valence but not specifically higher than arousal. '
          'Partial support for Davidson (1992). RQ2: mixed result.')
else:
    print('  FAA-only model does not reach significance for valence. '
          'Davidson (1992) effect undetected even in isolated FAA analysis. '
          'May require larger within-subject trial sets. RQ2: fail to reject null.')

# Plot
fig, ax = plt.subplots(figsize=(8, 4))
rng_jit = np.random.default_rng(99)
colors = ['steelblue', 'coral', 'seagreen']
for i, (t, col) in enumerate(zip(T_IDX, colors)):
    accs  = faa_accs[t]
    valid = accs[~np.isnan(accs)]
    jit   = rng_jit.uniform(-0.08, 0.08, len(valid))
    ax.scatter(np.full(len(valid), i) + jit, valid, alpha=0.45, s=22, color=col, label=t)
    ax.plot([i - 0.2, i + 0.2], [np.nanmean(valid)] * 2, color=col, lw=2.5)
ax.axhline(0.5, color='black', linestyle='--', lw=1, alpha=0.7, label='chance')
ax.set_xticks(range(3))
ax.set_xticklabels(list(T_IDX.keys()))
ax.set_ylabel('Per-subject FAA-only SVM accuracy (5-fold WS-CV)')
ax.set_title('RQ2: FAA-only within-subject SVM accuracy by affective target')
ax.legend()
plt.tight_layout()
plt.savefig(FIGS / 'rq2_faa_only.png', dpi=150)
plt.show()
```

## RQ3: Gap-Difference Paired Wilcoxon

**Problem in original analysis:** The generalization gap (WS accuracy minus LOSO accuracy) was reported as a mean with directional commentary, but no formal test was run on whether the gap magnitude differs significantly across affective dimensions. The claim that arousal shows a larger generalization penalty than valence was directional only.

**Fix:** Extract per-subject gap values (WS minus LOSO per subject, aligned by subject index). Run paired Wilcoxon on (arousal gap minus valence gap) and (dominance gap minus valence gap) with BH-FDR correction. Since both LOSO and WS-CV are indexed by the same subjects, the paired structure is natural.

**Note:** If `loso_results.pkl` stores only mean accuracy rather than per-fold values, the cell will print the available keys so you can adjust the extraction path.

```python
# Auto-detect per-fold LOSO accuracy and per-subject WS accuracy key paths
print('Inspecting pkl structures...')
print('  loso top-level:', list(loso.keys()) if isinstance(loso, dict) else type(loso))
print('  ws   top-level:', list(ws.keys())   if isinstance(ws,   dict) else type(ws))
if isinstance(loso, dict) and 'valence' in loso:
    print('  loso[valence]:', list(loso['valence'].keys()))
    if 'svm' in loso['valence']:
        print('  loso[valence][svm]:', list(loso['valence']['svm'].keys()))
if isinstance(ws, dict) and 'valence' in ws:
    print('  ws[valence]:', list(ws['valence'].keys()))
    if 'svm' in ws['valence']:
        print('  ws[valence][svm]:', list(ws['valence']['svm'].keys()))

# Candidate key paths for per-fold LOSO acc (strip target key; applied per target below)
LOSO_PATHS = [
    ['svm', 'eeg', 'fold_acc'],
    ['svm', 'eeg', 'per_fold_acc'],
    ['svm', 'eeg', 'fold_accs'],
    ['svm', 'fold_acc'],
    ['svm', 'per_fold_acc'],
]
WS_PATHS = [
    ['svm', 'per_subject_acc'],
    ['svm', 'subj_accs'],
    ['svm', 'subject_accs'],
    ['svm', 'acc'],
]

loso_path, ws_path = None, None
for path in LOSO_PATHS:
    if safe_extract(loso, 'valence', *path) is not None:
        loso_path = path
        print(f'LOSO path detected: valence -> {path}')
        break
for path in WS_PATHS:
    if safe_extract(ws, 'valence', *path) is not None:
        ws_path = path
        print(f'WS   path detected: valence -> {path}')
        break

if loso_path is None:
    print('WARNING: Could not detect LOSO per-fold acc path. '
          'Inspect loso dict above and set loso_path manually, e.g.:  loso_path = ["svm", "eeg", "fold_acc"]')
if ws_path is None:
    print('WARNING: Could not detect WS per-subject acc path. '
          'Inspect ws dict above and set ws_path manually, e.g.:  ws_path = ["svm", "per_subject_acc"]')
```

```python
# If auto-detection failed, override here:
# loso_path = ['svm', 'eeg', 'fold_acc']
# ws_path   = ['svm', 'per_subject_acc']

gaps = {}
if loso_path and ws_path:
    for t in T_IDX:
        lo_raw = safe_extract(loso, t, *loso_path)
        ws_raw = safe_extract(ws,   t, *ws_path)
        if lo_raw is None or ws_raw is None:
            print(f'  {t}: extraction returned None. Check paths.')
            continue
        lo = np.array(lo_raw, dtype=float)
        wa = np.array(ws_raw, dtype=float)
        n  = min(len(lo), len(wa))
        valid = ~(np.isnan(lo[:n]) | np.isnan(wa[:n]))
        gaps[t] = wa[:n][valid] - lo[:n][valid]
        print(f'  {t}: WS={wa[:n][valid].mean():.3f}  LOSO={lo[:n][valid].mean():.3f}  '
              f'gap={gaps[t].mean():.4f}  n_valid={valid.sum()}')

if len(gaps) == 3:
    comps = [
        ('arousal gap > valence gap',   gaps['arousal'],   gaps['valence']),
        ('dominance gap > valence gap', gaps['dominance'], gaps['valence']),
    ]
    raw_pvals, stats_wil = [], []
    for _, g_a, g_b in comps:
        n = min(len(g_a), len(g_b))
        s, p = wilcoxon(g_a[:n], g_b[:n], alternative='greater')
        raw_pvals.append(p)
        stats_wil.append(s)

    rej, pvals_fdr, _, _ = multipletests(raw_pvals, method='fdr_bh')

    print('\nRQ3: Gap-Difference Paired Wilcoxon (SVM EEG-only, BH-FDR corrected)')
    print(f'{"Comparison":<32} {"Stat":>8} {"p (raw)":>10} {"p (FDR)":>10} {"Decision":>14}')
    print('-' * 78)
    for i, (label, g_a, g_b) in enumerate(comps):
        dec = 'REJECT H0' if rej[i] else 'fail to reject'
        print(f'{label:<32} {stats_wil[i]:>8.2f} {raw_pvals[i]:>10.4f} {pvals_fdr[i]:>10.4f} {dec:>14}')

    print('\nConclusion:')
    if rej[0]:
        print('  Arousal gap is formally larger than valence gap. Cross-subject arousal '
              'encoding is more individually variable than valence encoding. RQ3: conclusive.')
    else:
        print('  Arousal vs valence gap difference is not statistically significant at N=32. '
              'Gap patterns are directional only. Power analysis: N~120-200 needed at these effect sizes.')
    if rej[1]:
        print('  Dominance gap is formally larger than valence gap. Dominance encoding '
              'is the most individually idiosyncratic of the three dimensions.')
    else:
        print('  Dominance vs valence gap difference is not statistically significant at N=32.')

    # Plot gap distributions
    fig, ax = plt.subplots(figsize=(8, 5))
    rng_jit = np.random.default_rng(7)
    colors  = ['steelblue', 'coral', 'seagreen']
    for i, (t, col) in enumerate(zip(T_IDX, colors)):
        if t not in gaps: continue
        g   = gaps[t]
        jit = rng_jit.uniform(-0.08, 0.08, len(g))
        ax.scatter(np.full(len(g), i) + jit, g, alpha=0.45, s=22, color=col)
        ax.boxplot(g, positions=[i], widths=0.35, patch_artist=True,
                   boxprops=dict(facecolor=col, alpha=0.3),
                   medianprops=dict(color='black', lw=2), showfliers=False)
    ax.axhline(0, color='black', linestyle='--', lw=1, alpha=0.6)
    ax.set_xticks(range(3))
    ax.set_xticklabels(list(T_IDX.keys()))
    ax.set_ylabel('Per-subject gap: WS acc minus LOSO acc  (SVM, EEG-only)')
    ax.set_title('RQ3: Generalization gap distribution by affective dimension')
    plt.tight_layout()
    plt.savefig(FIGS / 'rq3_gap_distribution.png', dpi=150)
    plt.show()

elif loso_path is None or ws_path is None:
    print('Gap computation skipped: set loso_path and ws_path in the cell above.')
```

## RQ4: DEAP is an Unsuitable Dataset for a Conclusive Answer

RQ4 asks whether EEG and peripheral channels contribute redundantly or non-redundantly across affective dimensions, with the prediction that peripheral channels are redundant with EEG for arousal (shared autonomic signature) and non-redundant for valence (centrally encoded). The directional pattern in the paper is consistent with this: peripheral channels outperform EEG for arousal under SVM (0.513 vs 0.504), and EEG outperforms peripheral for valence (0.525 vs 0.495).

**Why no analytical fix is possible within DEAP:**
The accuracy differences between feature configurations are 0.009 to 0.021 across 32 LOSO folds. A paired Wilcoxon test at N=32 requires effect sizes in the 0.05+ range to reach conventional power at alpha=0.05. A post-hoc power analysis for the observed 0.009-0.021 differences yields required N of approximately 120 to 200 subjects. DEAP provides 32. No permutation test, bootstrap, or resampling scheme can create statistical power that the sample size does not support. The peripheral features used (per-channel mean across 60 seconds) are also the weakest possible feature extraction, meaning the peripheral results represent a lower bound. Richer peripheral features (HRV, SCR peaks, respiratory rate) would likely shift the accuracy differences but would not resolve the fundamental power problem.

**Path to a conclusive answer:** Replication on a larger multimodal dataset. MAHNOB-HCI (27 subjects, richer peripheral channels) or pooling DEAP with DREAMER (23 subjects, same three affective dimensions) would increase N but likely remains underpowered. A purpose-designed study with N >= 100 subjects, richer peripheral feature extraction, and the same stimulus set across subjects would produce a conclusive answer. Within DEAP, RQ4 is structurally non-determinable and the directional evidence is the correct and final answer for this dataset.

## RQ5: Russell Null Simulation

**Problem in original analysis (Limitation L6):** All four valence-dominance importance correlation values fall within the 500-permutation null bounds (null p95 = 0.118-0.142), so the near-zero result is consistent with the PAD model but not statistically confirmatory. The null used (shuffled labels) tests against complete randomness. But the theoretical null for RQ5 is not randomness, it is: *what correlation would we expect if dominance were just noisy valence, as Russell's circumplex reduction implies?*

**Fix:** Simulate the Russell alternative null. For a range of label noise levels (flip_rate from 0% to 50%), create perturbed valence binary labels (randomly flipping flip_rate fraction of labels), run lean within-subject RF, and compute Spearman correlation between the pre-computed true valence importance vector and the perturbed-valence importance vector. At 0% flip: this is valence correlated with itself (expected high correlation). At 50% flip: labels are essentially random (expected near-zero correlation). The resulting distributions show what the importance correlation looks like under Russell's hypothesis at each noise level.

**Formal test:** At each flip_rate, is our observed valence-dominance importance correlation below the p5 of the Russell null distribution? If yes at low flip_rates, the Russell null is rejected even for mildly noisy shared-feature scenarios. The crossover flip_rate where the observed value stops being below p5 tells us: how much noise would dominance labels need to contain before our data becomes consistent with Russell's prediction?

```python
# Pre-binarize valence labels per subject (per-subject median, for simulation use)
# This approximates the fold-specific median used in the main pipeline
y_val_bin = np.zeros(len(y), dtype=int)
for s in SUBJECTS:
    mask = sids == s
    thresh = np.median(y[mask, 0])
    y_val_bin[mask] = (y[mask, 0] > thresh).astype(int)

# Reference importance vector: pre-computed WS-RF valence importance
ref_imp = imp_rf['valence']   # (131,)

def ws_rf_importance_perturbed(X_full, y_bin, flip_rate, rng, n_est=50, n_reps=3):
    """Within-subject RF with flip_rate fraction of binary labels flipped.
    Returns mean 131-dim importance vector across all subjects and folds.
    """
    all_imp = []
    for s in SUBJECTS:
        mask = sids == s
        Xs   = X_full[mask]
        ys   = y_bin[mask].copy()
        flip = rng.random(len(ys)) < flip_rate
        ys[flip] = 1 - ys[flip]
        if len(np.unique(ys)) < 2:
            continue
        skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
        subj_imp = []
        for tri, tei in skf.split(Xs, ys):
            Xtr, Xte = Xs[tri], Xs[tei]
            ytr, yte = ys[tri], ys[tei]
            if len(np.unique(ytr)) < 2:
                continue
            sc  = StandardScaler()
            Xtr = sc.fit_transform(Xtr)
            Xte = sc.transform(Xte)
            rf  = RandomForestClassifier(n_estimators=n_est, random_state=42, n_jobs=1)
            rf.fit(Xtr, ytr)
            pi  = permutation_importance(rf, Xte, yte, n_repeats=n_reps, random_state=42)
            subj_imp.append(pi.importances_mean)
        if subj_imp:
            all_imp.append(np.mean(subj_imp, axis=0))
    return np.mean(all_imp, axis=0) if all_imp else None

print('Setup complete. Reference importance shape:', ref_imp.shape)
```

```python
# Russell null simulation (checkpointed)
# Runtime estimate: 100 sims x 32 subjects x 5 folds x (50-est RF + 3-rep perm) ~ 10-20 minutes
FLIP_RATES = [0.0, 0.10, 0.20, 0.30, 0.40, 0.50]
N_SIM      = 20   # simulations per flip rate; increase to 50 for tighter null intervals
RQ5_PATH   = RES / 'rq5_russell_null.pkl'

if RQ5_PATH.exists():
    with open(RQ5_PATH, 'rb') as f:
        rq5_null = pickle.load(f)
    print('Loaded Russell null from checkpoint.')
    print({fr: len(v) for fr, v in rq5_null.items()})
else:
    rq5_null = {fr: [] for fr in FLIP_RATES}
    rng5 = np.random.default_rng(1)
    for fr in FLIP_RATES:
        needed = N_SIM - len(rq5_null[fr])
        if needed <= 0:
            continue
        print(f'flip_rate={fr:.2f} ({needed} sims):', end=' ', flush=True)
        for _ in range(needed):
            sim_imp = ws_rf_importance_perturbed(X, y_val_bin, fr, rng5)
            if sim_imp is not None:
                rho = spearmanr(ref_imp, sim_imp)[0]
                rq5_null[fr].append(rho)
            print('.', end='', flush=True)
        rq5_null[fr] = np.array(rq5_null[fr])
        p5 = np.percentile(rq5_null[fr], 5) if len(rq5_null[fr]) > 0 else float('nan')
        print(f'  mean={np.mean(rq5_null[fr]):.3f}  p5={p5:.3f}')
    with open(RQ5_PATH, 'wb') as f:
        pickle.dump(rq5_null, f)
    print('Saved to results/rq5_russell_null.pkl')
```

```python
# Observed valence-dominance importance correlations (re-computed from loaded vectors)
obs_vd_rf  = spearmanr(imp_rf['valence'],  imp_rf['dominance'])[0]
obs_vd_svm = spearmanr(imp_svm['valence'], imp_svm['dominance'])[0]
print(f'Observed WS-RF  valence-dominance Spearman: {obs_vd_rf:.4f}')
print(f'Observed WS-SVM valence-dominance Spearman: {obs_vd_svm:.4f}')

print(f'\nRussell Null: is observed WS-RF correlation below null p5 at each flip rate?')
print(f'(obs < p5 = observed inconsistent with Russell null at that noise level)')
print(f'{"Flip Rate":<12} {"N sims":>8} {"Null mean":>12} {"Null p5":>10} '
      f'{"Obs RF":>10} {"Obs < p5":>10}')
print('-' * 66)

crossover_fr = None
for fr in FLIP_RATES:
    null = np.array(rq5_null[fr])
    if len(null) == 0:
        continue
    p5     = np.percentile(null, 5)
    below  = obs_vd_rf < p5
    if not below and crossover_fr is None:
        crossover_fr = fr
    print(f'{fr:<12.2f} {len(null):>8} {null.mean():>12.4f} {p5:>10.4f} '
          f'{obs_vd_rf:>10.4f} {str(below):>10}')

print('\nConclusion:')
if crossover_fr is None:
    print('  Observed correlation remains below Russell null p5 across ALL flip rates (0% to 50%).')
    print('  Russell null is rejected even at maximal label noise. Strong computational support for PAD.')
elif crossover_fr == 0.0:
    print('  Observed correlation is NOT below Russell null p5 even at 0% flip rate.')
    print('  This is unexpected. Check that imp_rf vectors are correctly loaded.')
else:
    pct = int(crossover_fr * 100)
    print(f'  Observed correlation falls below Russell null p5 for flip_rate < {crossover_fr:.2f}.')
    print(f'  Russell null is rejected unless dominance labels contain >= {pct}% noise relative to valence.')
    if crossover_fr <= 0.20:
        print(f'  A crossover at {pct}% is low: our data is inconsistent with even a weakly noisy '
              f'shared-feature model. Good computational evidence for PAD-model orthogonality.')
    elif crossover_fr <= 0.40:
        print(f'  A crossover at {pct}% is moderate: Russell null rejected for clean or mildly noisy '
              f'shared-feature models. Exploratory support for PAD at this sample size.')
    else:
        print(f'  A crossover at {pct}% (near-random noise): our observation is consistent with '
              f'a highly noisy Russell null. Limited discriminating power at this sample size.')

# Violin plot of Russell null distributions with observed values overlaid
fig, ax = plt.subplots(figsize=(10, 5))

positions = [i for i, fr in enumerate(FLIP_RATES) if len(rq5_null[fr]) > 0]
data      = [np.array(rq5_null[fr]) for fr in FLIP_RATES if len(rq5_null[fr]) > 0]
labels    = [f'{fr:.0%}' for fr in FLIP_RATES if len(rq5_null[fr]) > 0]

parts = ax.violinplot(data, positions=positions, showmedians=True, showextrema=True)
for pc in parts['bodies']:
    pc.set_facecolor('lightblue')
    pc.set_alpha(0.6)

ax.axhline(obs_vd_rf,  color='crimson',   linestyle='--', lw=1.8,
           label=f'WS-RF obs ({obs_vd_rf:.3f})')
ax.axhline(obs_vd_svm, color='darkorange', linestyle=':',  lw=1.8,
           label=f'WS-SVM obs ({obs_vd_svm:.3f})')
ax.axhline(0, color='gray', linestyle='-', lw=0.8, alpha=0.4)

ax.set_xticks(positions)
ax.set_xticklabels(labels)
ax.set_xlabel('Fraction of valence labels flipped (Russell noise level)', fontsize=11)
ax.set_ylabel('Spearman correlation with true valence importance', fontsize=11)
ax.set_title('RQ5: Russell Null Simulation\n'
             'Importance correlation expected if dominance = noisy valence vs observed valence-dominance', fontsize=10)
ax.legend(fontsize=10)
plt.tight_layout()
plt.savefig(FIGS / 'rq5_russell_null.png', dpi=150)
plt.show()
```
