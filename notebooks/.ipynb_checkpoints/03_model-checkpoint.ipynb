# 03_model.ipynb — Model Training, Within-Subject CV, and Null Distribution


| Section | Runtime | Purpose |
|---|---|---|
| 1. LOSO main loop | 8-15 min | Cross-subject RF + SVM, all configs and targets |
| 2. Within-subject CV | 15-30 min | Per-subject 5-fold CV (RQ1/2 importance, RQ3 gap baseline) |
| 3. SVM importance (LOSO) | 30-50 min | Cross-condition RQ5 corroboration |
| 4. MI robustness | <1 min | Importance correlation on MI-selected feature subset |
| 5. Save all | — | All .pkl and .npy outputs |
| 6. Null distribution | 30-50 min | Shuffled-label null for RQ1 and RQ5 (checkpointed) |

**Run once. Do not re-run unless results are corrupted.**

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import pickle
from pathlib import Path

from models import (
    run_loso, aggregate_importance,
    importance_correlation, shuffled_label_null,
    paired_accuracy_tests, fdr_correction,
    apply_fdr_to_correlations,
    run_loso_svm_importance, run_within_subject,
    sign_test_ws, mi_importance_correlation, corr_on_selected,
    TARGETS, BAND_NAMES, CONFIGS, CORR_KEYS,
)

Path('../results').mkdir(exist_ok=True)

X_features            = np.load('../results/X_features.npy').astype(np.float32)
X_peripheral_features = np.load('../results/X_peripheral_features.npy').astype(np.float32)
y                     = np.load('../results/y.npy')
subject_ids           = np.load('../results/subject_ids.npy')

print(f'X_features:            {X_features.shape}  dtype={X_features.dtype}')
print(f'X_peripheral_features: {X_peripheral_features.shape}')
print(f'Subjects: {len(np.unique(subject_ids))}  |  Trials: {len(y)}')
```

## 1. LOSO main loop
32 folds, 3 targets, 3 feature configs, 2 classifiers.
Includes RF permutation importance (test fold only) and band ablation.
Expected runtime: 8-15 min.

```python
results = run_loso(
    X_eeg            = X_features,
    X_peripheral     = X_peripheral_features,
    y                = y,
    subject_ids      = subject_ids,
    n_perm_repeats   = 20,
    rf_n_estimators  = 150,
    n_jobs           = -3,
)
print('LOSO complete.')
```

```python
# Accuracy table — flag anything below 0.50
print(f'{"Target":<12} {"Config":<12} {"Model":<6} {"Acc":>6} {"Std":>6} {"F1":>6}')
print('-' * 52)
for target in TARGETS:
    for cfg in CONFIGS:
        for model in ('svm', 'rf'):
            acc = results['metrics'][target][cfg][model]['acc']
            f1  = results['metrics'][target][cfg][model]['f1']
            flag = ' <BELOW-CHANCE-BUG' if np.mean(acc) < 0.48 else ''
            print(f'{target:<12} {cfg:<12} {model:<6} '
                  f'{np.mean(acc):>6.3f} {np.std(acc):>6.3f} {np.mean(f1):>6.3f}{flag}')

# Class balance flags
any_flagged = False
for target in TARGETS:
    flagged = results['flagged_folds'][target]
    if flagged:
        print(f'\n[FLAG] {target}: imbalance >65/35 in subjects {flagged}')
        any_flagged = True
if not any_flagged:
    print('\nNo flagged folds — all class ratios within 65/35.')
```

```python
# Band ablation summary (primary RQ1 cross-subject evidence)
print('Band ablation LOSO RF accuracy:')
print(f'{"Band":<8}', '  '.join(f'{t:>10}' for t in TARGETS))
print('-' * 44)
for band in BAND_NAMES:
    row = '  '.join(
        f'{np.mean(results["ablation"][t][band]["acc"]):>10.3f}'
        for t in TARGETS
    )
    print(f'{band:<8}  {row}')
print('\nFull LOSO RF reference:')
for t in TARGETS:
    print(f'  {t}: {np.mean(results["metrics"][t]["eeg"]["rf"]["acc"]):.3f}')
```

```python
# Paired Wilcoxon on LOSO fold accuracy (RQ4 directional support)
paired    = paired_accuracy_tests(results['fold_acc'])
pair_pvals = {k: v['pvalue'] for k, v in paired.items()}
pair_fdr   = fdr_correction(pair_pvals, alpha=0.05)

print('Paired Wilcoxon on LOSO fold accuracy (BH-FDR corrected):')
print(f'{"Comparison":<45} {"mean_diff":>10} {"p_fdr":>8} {"sig":>5}')
print('-' * 72)
for key, vals in paired.items():
    fdr_p = pair_fdr[key]['pvalue_fdr']
    sig   = '*' if fdr_p < 0.05 else ''
    print(f'{key:<45} {vals["mean_diff"]:>10.4f} {fdr_p:>8.4f} {sig:>5}')
print('* = significant after BH-FDR; positive = EEG > peripheral for eeg_vs_peripheral')
```

## 2. Within-subject 5-fold CV
Eliminates cross-subject distribution shift.
Provides: per-subject accuracy (RQ3 gap baseline), sign test, band/FAA importance (RQ1/2), importance correlation (RQ5 within-subject).
Expected runtime: 15-30 min.

```python
ws = run_within_subject(
    X_eeg           = X_features,
    y               = y,
    subject_ids     = subject_ids,
    n_splits        = 5,
    n_perm_repeats  = 20,
    rf_n_estimators = 150,
    n_jobs          = -3,
)
print('Within-subject CV complete.')
```

```python
# Usable subjects per target
for target in TARGETS:
    for model in ('svm', 'rf'):
        acc_arr = np.array(ws['metrics'][target][model]['acc'])
        n_valid = int(np.sum(~np.isnan(acc_arr)))
        mean_acc = float(np.nanmean(acc_arr))
        print(f'{target:<12} {model:<5} valid={n_valid}/32  acc={mean_acc:.3f}')
```

```python
# Sign test (formally tests individual-level above-chance accuracy)
sign = ws['sign_test']
print('Sign test — subjects exceeding 0.50 accuracy (one-sided binomial):')
print(f'{"Target + Model":<25} {"n_above/n_valid":>16} {"p-value":>10}  sig')
print('-' * 60)
for key, vals in sign.items():
    sig = '*' if vals['pvalue'] < 0.05 else ''
    print(f'{key:<25} {vals["n_above"]:>6}/{vals["n_valid"]:<9} '
          f'{vals["pvalue"]:>10.4f}  {sig}')
print('* = p < 0.05 (H0: p_above=0.5)')
```

```python
# Within-subject band-level importance (both models)
for model in ('svm', 'rf'):
    print(f'\nWithin-subject {model.upper()} band importance:')
    print(f'{"Band":<8}', '  '.join(f'{t:>10}' for t in TARGETS))
    print('-' * 44)
    for b, band in enumerate(BAND_NAMES):
        row = '  '.join(
            f'{ws["importance_agg"][t][model]["band"][b]:>10.5f}'
            for t in TARGETS
        )
        print(f'{band:<8}  {row}')
print('\nAll-negative values = near-chance classifiers (expected at this evaluation scale).')
print('Primary RQ1 evidence comes from band ablation (above), not permutation importance.')
```

```python
# Importance correlation across WS-SVM, WS-RF (RQ5 within-subject)
print('Within-subject importance correlation:')
print(f'{"Pair":<30} {"WS-SVM":>8} {"WS-RF":>8}')
print('-' * 50)
for key in CORR_KEYS:
    ws_svm = ws['importance_corr']['svm'][key]['rho']
    ws_rf  = ws['importance_corr']['rf'][key]['rho']
    print(f'{key:<30} {ws_svm:>8.3f} {ws_rf:>8.3f}')
print('\nValence-dominance near zero across both = primary RQ5 robustness evidence.')
```

## 3. SVM permutation importance (LOSO)
Used only for cross-condition RQ5 corroboration.
Band-level SVM LOSO values are excluded from all figures (all-negative at near-chance).
Expected runtime: 30-50 min.

```python
svm_imp = run_loso_svm_importance(
    X_eeg          = X_features,
    y              = y,
    subject_ids    = subject_ids,
    n_perm_repeats = 20,
    n_jobs         = -3,
)
print('SVM importance loop complete.')
```

```python
# LOSO SVM vs RF importance correlation (RQ5 cross-condition table)
# loso_rf_corr is computed after LOSO — stored in results via importance_correlation()
rf_corr = importance_correlation(results['importance_mean'])

print('Pairwise Spearman importance correlation — all conditions:')
print(f'{"Pair":<30} {"LOSO-RF":>8} {"LOSO-SVM":>9} {"WS-SVM":>8} {"WS-RF":>8}')
print('-' * 68)
for key in CORR_KEYS:
    lo_rf  = rf_corr[key]['rho']
    lo_svm = svm_imp['importance_corr'][key]['rho']
    ws_svm = ws['importance_corr']['svm'][key]['rho']
    ws_rf  = ws['importance_corr']['rf'][key]['rho']
    print(f'{key:<30} {lo_rf:>8.3f} {lo_svm:>9.3f} {ws_svm:>8.3f} {ws_rf:>8.3f}')
print('\nValence-dominance near zero across all 4 = primary RQ5 finding.')
print('Arousal-dominance sign-inconsistent across models = inconclusive (as documented).')
```

## 4. MI importance robustness (fast)

```python
# Does near-zero valence-dominance correlation hold when restricted to
# the top-50 MI-selected features per target? Rules out noise-feature confound.
mi_info = mi_importance_correlation(X_features, y, subject_ids, k=50)
print(f'MI selection: {mi_info["n_selected"]} features (union of top-50 per target, from 131)')

ws_svm_mi = corr_on_selected(
    {t: ws['importance_mean'][t]['svm'] for t in TARGETS},
    mi_info['selected_indices']
)
ws_rf_mi = corr_on_selected(
    {t: ws['importance_mean'][t]['rf'] for t in TARGETS},
    mi_info['selected_indices']
)

print('\nImportance correlation: full 131 features vs MI-selected subset:')
print(f'{"Pair":<30} {"WS-SVM full":>12} {"WS-SVM MI":>11} {"WS-RF full":>11} {"WS-RF MI":>10}')
print('-' * 78)
for key in CORR_KEYS:
    svm_f  = ws['importance_corr']['svm'][key]['rho']
    rf_f   = ws['importance_corr']['rf'][key]['rho']
    svm_mi = ws_svm_mi[key]['rho']
    rf_mi  = ws_rf_mi[key]['rho']
    print(f'{key:<30} {svm_f:>12.3f} {svm_mi:>11.3f} {rf_f:>11.3f} {rf_mi:>10.3f}')
print('\nIf valence-dominance stable on MI subset → inductive-bias confound ruled out.')
```

## 5. Save all results

```python
# Aggregate RF importance (for figures)
importance_agg = aggregate_importance(results['importance_mean'])

with open('../results/loso_results.pkl',   'wb') as f: pickle.dump(results,        f)
with open('../results/importance_agg.pkl', 'wb') as f: pickle.dump(importance_agg, f)
with open('../results/importance_corr.pkl','wb') as f: pickle.dump(rf_corr,        f)
with open('../results/svm_importance.pkl', 'wb') as f: pickle.dump(svm_imp,        f)
with open('../results/within_subject.pkl', 'wb') as f: pickle.dump(ws,             f)

for target in TARGETS:
    np.save(f'../results/importance_mean_{target}.npy',
            results['importance_mean'][target])
    np.save(f'../results/svm_importance_mean_{target}.npy',
            svm_imp['importance_mean'][target])

np.save('../results/mi_selected_indices.npy',
        np.array(mi_info['selected_indices']))

sizes = {p.name: f'{p.stat().st_size/1e6:.1f} MB'
         for p in Path('../results').glob('*.pkl')}
for name, sz in sizes.items():
    print(f'  {name}: {sz}')
print('\nAll results saved. Starting null distribution next.')
```

## 6. Null distribution (checkpointed)
Batched with checkpoint saving. Safe to interrupt — resumes from last completed batch.

Settings: 500 permutations, batch_size=50, lean RF (50 trees, 3 perm repeats).
Expected runtime: 30-50 min across 10 batches.

```python
null_results = shuffled_label_null(
    X_eeg        = X_features,
    X_peripheral = X_peripheral_features,
    y            = y,
    subject_ids  = subject_ids,
)
print('Null distribution complete.')
```

```python
# Observed vs null comparison
print('Null accuracy (EEG RF, LOSO) — mean | p5 | p95 | obs:')
for target in TARGETS:
    s   = null_results['accuracy'][target]
    obs = np.mean(results['metrics'][target]['eeg']['rf']['acc'])
    sig = 'ABOVE NULL' if obs > s['p95'] else 'within null'
    print(f'  {target:<12} null={s["mean"]:.3f} p5={s["p5"]:.3f} '
          f'p95={s["p95"]:.3f} n={s["n"]}  obs={obs:.3f}  -> {sig}')

print('\nImportance correlation vs null (obs | null p5 | null p95):')
for key in CORR_KEYS:
    s   = null_results['correlation'][key]
    obs = rf_corr[key]['rho']
    flag = ('ABOVE p95' if obs > s['p95'] else
            'BELOW p5'  if obs < s['p5']  else 'within null')
    print(f'  {key:<30} p5={s["p5"]:.3f} p95={s["p95"]:.3f}  obs={obs:.3f}  -> {flag}')
```

```python
with open('../results/null_results.pkl', 'wb') as f:
    pickle.dump(null_results, f)
print('Saved: null_results.pkl')
print('All 03_model outputs saved. Ready for 04_results.ipynb.')
```
